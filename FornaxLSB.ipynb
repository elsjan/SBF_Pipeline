```python
from astropy.io import fits
import matplotlib.pyplot as plt
import numpy as np
from euclid_sbf_pipeline import MainPipeline
```

```python
def getpsf(file_path):
    psf_data, psf_header = fits.getdata(file_path, ext=1, header=True) 
    plt.imshow(psf_data[0][0][0], origin="lower", cmap="viridis")
    plt.colorbar(label=f"PSF intensity")
    plt.title(f"PSF from {file_path}")
    plt.show()    
    psf = psf_data[0][0][0]
    psf = psf / np.sum(psf)  # Normalize the PSF
    return psf
```

```python
path_psf_144 = "psfex-sex-files/starsFCC144LSB.psf"
path_psf_146 = "psfex-sex-files/starsFCC146LSB.psf"
path_psf_167 = "psfex-sex-files/starsFCC167LSB.psf"
path_psf_168 = "psfex-sex-files/starsFCC168LSB.psf"
path_psf_184 = "psfex-sex-files/starsFCC184LSB.psf"

psf144 = getpsf(path_psf_144)
psf146 = getpsf(path_psf_146)
psf167 = getpsf(path_psf_167)
psf168 = getpsf(path_psf_168)
psf184 = getpsf(path_psf_184)
```

```python
data_path = 'Euclid/data/FornaxLSB/FCC144/flt'
file_path = 'Euclid/data/FornaxLSB/FCC144/return_files'
image_path = 'Euclid/data/FornaxLSB/FCC144/images'
import os
try:
    os.makedirs(file_path)
except FileExistsError:
    # directory already exists
    pass
try:
    os.makedirs(image_path)
except FileExistsError:
    # directory already exists
    pass


data144, mask_model144, mask_combined144, nri144, image_ps144, expected_ps144, sbf144, noise144, sbfmag144 = MainPipeline(data_path, image_path=image_path, make_plots=True, psf=psf144,background_estimation=True)
```
